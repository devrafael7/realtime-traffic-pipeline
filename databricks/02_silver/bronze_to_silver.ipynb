```sql
%sql

CREATE OR REPLACE TABLE traffic_catalog.silver.silver_traffic
USING DELTA
AS

SELECT
    CAST(get_json_object(message, '$.road_id') AS INT) AS road_id,

    get_json_object(message, '$.road_name') AS road_name,

    get_json_object(message, '$.road_code') AS road_code,

    CAST(get_json_object(message, '$.latitude') AS DOUBLE) AS latitude,

    CAST(get_json_object(message, '$.longitude') AS DOUBLE) AS longitude,

    CAST(get_json_object(message, '$.current_speed') AS DOUBLE) AS current_speed,

    CAST(get_json_object(message, '$.free_flow_speed') AS DOUBLE) AS free_flow_speed,

    CAST(get_json_object(message, '$.current_travel_time') AS INT) AS current_travel_time,

    CAST(get_json_object(message, '$.free_flow_travel_time') AS INT) AS free_flow_travel_time,

    CAST(get_json_object(message, '$.confidence') AS DOUBLE) AS confidence,

    CAST(get_json_object(message, '$.road_closure') AS BOOLEAN) AS road_closure,

    FROM_UNIXTIME(
        CAST(get_json_object(message, '$.timestamp') AS DOUBLE)
    ) AS event_timestamp,

    kafka_timestamp,

    ingestion_timestamp,

    topic,

    partition,

    offset

FROM traffic_catalog.bronze.bronze_traffic;
```
