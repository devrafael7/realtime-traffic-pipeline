```python
%sql

CREATE OR REPLACE TABLE traffic_catalog.gold.gold_road_congestion_ranking
USING DELTA
AS

SELECT
    road_id,
    road_name,
    road_code,

    ROUND(AVG(current_speed), 2) AS avg_speed,

    ROUND(AVG(free_flow_speed), 2) AS avg_free_flow_speed,

    ROUND(
        (
            1 - AVG(current_speed) /
            NULLIF(AVG(free_flow_speed), 0)
        ) * 100,
        2
    ) AS avg_speed_reduction,

    COUNT(*) AS total_measurements

FROM traffic_catalog.silver.silver_traffic

GROUP BY
    road_id,
    road_name,
    road_code

ORDER BY
    avg_speed_reduction DESC;
```
