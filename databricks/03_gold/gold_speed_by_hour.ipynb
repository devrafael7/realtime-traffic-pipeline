```python
%sql

CREATE OR REPLACE TABLE traffic_catalog.gold.gold_speed_by_hour
USING DELTA
AS

SELECT
    road_id,
    road_name,

    HOUR(event_timestamp) AS hour,

    ROUND(AVG(current_speed), 2) AS avg_current_speed,

    ROUND(AVG(free_flow_speed), 2) AS avg_free_flow_speed,

    ROUND(
        (
            1 - AVG(current_speed) /
            NULLIF(AVG(free_flow_speed), 0)
        ) * 100,
        2
    ) AS speed_reduction_percentage,

    COUNT(*) AS measurement_count

FROM traffic_catalog.silver.silver_traffic

GROUP BY
    road_id,
    road_name,
    HOUR(event_timestamp);
```
