```python
%sql

CREATE OR REPLACE TABLE traffic_catalog.gold.gold_road_closures
USING DELTA
AS

SELECT
    road_id,
    road_name,
    road_code,

    COUNT(*) AS total_measurements,

    SUM(
        CASE
            WHEN road_closure = true THEN 1
            ELSE 0
        END
    ) AS closure_events,

    ROUND(
        SUM(
            CASE
                WHEN road_closure = true THEN 1
                ELSE 0
            END
        ) * 100.0 / COUNT(*),
        2
    ) AS closure_percentage,

    MIN(event_timestamp) AS first_measurement,

    MAX(event_timestamp) AS last_measurement

FROM traffic_catalog.silver.silver_traffic

GROUP BY
    road_id,
    road_name,
    road_code;
```
